## Step 10 — The chapter's `diagnose_request()`, read together (8 minutes)

A **function** is lines with a name, which run each time it's called with `( )`. `def` defines one; its **parameters**, here `url` and `headers`, are filled in by each call. The chapter's `diagnose_request()` gathers what Steps 8 and 9 printed. It's given without the chapter's `try`/`except` and `timeout=`, which keep a dead address from stopping the notebook. Find the lines you typed in it, then run it on our three addresses. See ["Diagnosing HTTP Errors"](https://cuinfoscience.github.io/Web-Data-Science-Book/ch-05-protocols.html#diagnosing-http-errors).

```python
# Given: the chapter's diagnose_request(), without its try/except and timeout
def diagnose_request(url, headers=None):
    """Print diagnostic information about an HTTP request."""
    response = requests.get(url, headers=headers)
    print(f"URL: {url}")
    print(f"Status: {response.status_code}")
    print(f"Content-Type: {response.headers.get('Content-Type', 'not specified')}")
    print(f"Server: {response.headers.get('Server', 'not specified')}")
    print(f"Response size: {len(response.content):,} bytes")
    if response.history:
        print(f"Redirected from: {response.history[0].url}")
    return response
```

```python
# Given: the function on the three addresses we requested by hand, a second apart
for url in [PAGE, missing_url, alias_url]:
    diagnose_request(url, headers=HEADERS)
    print()
    time.sleep(1)
```
