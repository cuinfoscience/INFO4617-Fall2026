## Step 3 — From a notebook to a script (7 minutes)

A collector is the cells above, saved as a script that a computer can run without us: `collector.py`, in the template. Read it together, and find our request in it. Then find four things a notebook didn't need:
- **constants** at the top (`URL`, `HEADERS`, `FILE`, `FIELDS`, `LAST_DAY`), so each is changed in one place;
- a CSV opened to **append** (`"a"`), with its header written only when the file is new ([`csv.DictWriter`](https://docs.python.org/3/library/csv.html#csv.DictWriter));
- [`raise_for_status()`](https://requests.readthedocs.io/en/stable/api/#requests.Response.raise_for_status), which stops the script with an error if the NWS didn't answer 200. Our labs never use it; a script that runs while nobody watches has to fail where someone will see it;
- [`if __name__ == "__main__":`](https://docs.python.org/3/library/__main__.html#idiomatic-usage), which runs `main()` only when the file is run as a script. After December 11, `main()` stops before it asks.

```python
# Given: read the template's collector.py, the file in the zip, and print it
# Uses template and HEADERS, from Step 0's cell
collector_response = requests.get(template + "collector.py", headers=HEADERS)
if collector_response.status_code != 200:
    raise RuntimeError(f"Got {collector_response.status_code}, not 200: "
                       f"{collector_response.text[:200]}")
collector = collector_response.text
print(collector)
```

## Step 10 — Is it still running? A freshness check (6 minutes)

A collector that stops doesn't say so: its file just stops growing. A **freshness check** asks how old the newest row is. How old is too old? The schedule says an hour, but GitHub ran this collector hours apart, so a given cell prints the hours from each run to the next ([`.diff()`](https://pandas.pydata.org/docs/reference/api/pandas.Series.diff.html) subtracts the row before): the longest gap is our yardstick. [`pd.Timestamp.now(tz="UTC")`](https://pandas.pydata.org/docs/reference/api/pandas.Timestamp.now.html) is the time now, in UTC, like `collected_at`; without `tz="UTC"`, the subtraction stops with a `TypeError`. Check the course's series, then your own, then the forecast itself: rows can keep coming while `update_time` stands still. For your own, a 404 means no run has finished, or the address has a typo: look at your Actions tab. No repository? Skip that cell.

```python
# Given: the hours from each run to the next, the longest of those gaps, and how many runs each day (UTC) had
# Uses forecasts, from Step 7
run_times = pd.to_datetime(forecasts["collected_at"].drop_duplicates())
gaps = run_times.diff().dropna()
gap_hours = gaps.dt.total_seconds() / 3600
print(gap_hours.round(1).tolist())
print("longest gap:", round(gap_hours.max(), 1), "hours")
print(run_times.dt.strftime("%b %d").value_counts(sort=False).to_dict())
```

## Step 9 — One hour's forecast, as it came closer (9 minutes)

Each run forecasts every hour of the next six and a half days, so a single hour is forecast again by every run until it comes: about 156 times if a run came every hour, and a few times a day as GitHub runs ours. Take one: noon on Wednesday, November 18, during Wednesday's class. Pick its rows with `==`, equal to, as text. Its **lead time** in each run is how far ahead the forecast was made: the hour's `start_time` minus the run's `collected_at`. A given cell works it out in hours, with [`.dt.total_seconds()`](https://pandas.pydata.org/docs/reference/api/pandas.Series.dt.total_seconds.html), and plots the forecast temperature against it, far ahead on the left, as steps, with [`ax.step(..., where="post")`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.step.html), as the chapter's afternoon chart does: a run's forecast stands until the next run, and several runs found the same one, so a sloping line would draw changes no run saw. The two columns have different offsets, UTC and Mountain time, hence `utc=True`; matching `-07:00` as text works only until the clocks change in March.

```python
# noon_rows: the rows whose "start_time" is "2026-11-18T12:00:00-07:00"; print how many
# Your code here
```

```python
# Given: each row's lead time in hours (start_time minus collected_at); plot the forecast temperature against it
# Uses noon_rows, from the cell above, and plt, from Step 7
lead = pd.to_datetime(noon_rows["start_time"], utc=True) - pd.to_datetime(noon_rows["collected_at"], utc=True)
lead_hours = lead.dt.total_seconds() / 3600
fig, ax = plt.subplots(figsize=(8, 4))
# One stepped line; a step holds each run's forecast until the next run's
ax.step(lead_hours, noon_rows["temperature"], where="post")
ax.invert_xaxis()  # far ahead on the left, the hour itself on the right
ax.set_title("Boulder's forecast for noon, Wednesday, November 18, run by run")
ax.set_xlabel("Lead time: hours between the run and noon on November 18")
ax.set_ylabel("Forecast temperature (°F)")
plt.show()
# Several runs found the same forecast: how many distinct forecasts of noon there were
print(noon_rows["update_time"].nunique(), "forecasts of noon in", len(noon_rows), "runs")
```
