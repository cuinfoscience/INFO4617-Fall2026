```python
# director_row: infobox_rows[2]; print its find("th").get_text(strip=True),
# then its find("td").get_text(" ", strip=True)
# Your code here
```

```python
# Given: the chapter's loop, as a function: your last cell, for every row of a table. A row with
# no <th> or no <td> gets None from find(), and "if label and value" skips it
def read_infobox(table):
    facts = {}
    for row in table.find_all("tr"):
        label = row.find("th")
        value = row.find("td")
        if label and value:
            key = label.get_text(strip=True)
            facts[key] = value.get_text(" ", strip=True)
    return facts
```

```python
# Given: call it on the English infobox; then how many facts, the director, and every label
facts = read_infobox(infobox)
print(len(facts), facts["Directed by"])
print(list(facts))
```

## Step 12 — One function, several ceremonies (8 minutes)

We want more years, so the given cell gathers Step 11's fetch, area and loop into the chapter's function, `parse_oscar_ceremony(year)`, with the year in the URL and a `"year"` column. Find your lines in it. Run it on one year by hand; then the given loop runs three, and [sleeps a second](https://docs.python.org/3/library/time.html#time.sleep) between them. The chapter's version fetches through week 2's `responsible_get()`, which returns `None` when a request fails; ours assumes every request works. See the chapter's ["Abstracting into Functions"](https://cuinfoscience.github.io/Web-Data-Science-Book/ch-06-static-pages.html#abstracting-into-functions).

```python
# Given: Step 11 as a function of the year. It returns one row per honoree
def parse_oscar_ceremony(year):
    page_url = "https://www.oscars.org/oscars/ceremonies/" + str(year)
    response = requests.get(page_url, headers=HEADERS)
    soup = BeautifulSoup(response.text, "html.parser")
    results = []
    for card in soup.find("div", class_=AREA).find_all("div", class_=CATEGORY):
        category_name = card.find(class_=CATEGORY_NAME).get_text(strip=True)
        for honoree in card.find_all("div", class_=HONOREE):
            honoree_film = honoree.find(class_=FILM)
            film = None
            if honoree_film:
                film = honoree_film.get_text(strip=True)
            results.append({
                "category": category_name,
                "honoree_type": honoree.find(class_=HONOREE_TYPE).get_text(strip=True),
                "entity": honoree.find(class_=ENTITY).get_text(strip=True),
                "film": film,
                "year": year,
            })
    return pd.DataFrame(results)
```

```python
# Given: three ceremonies, a second apart, in one table; then each year's categories and honorees
frames = []
for year in [2024, 2025, 2026]:
    frames.append(parse_oscar_ceremony(year))
    time.sleep(1)
oscars = pd.concat(frames, ignore_index=True)
oscars.groupby("year").agg(categories=("category", "nunique"), honorees=("category", "size"))
```
